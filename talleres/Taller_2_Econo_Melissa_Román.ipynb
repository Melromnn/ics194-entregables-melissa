3. (15 puntos) Realiza un test de hipotesis para evaluar si las dos variables que se agregaron son
significativas para el modelo simultaneamente. Use significancia del 5%.

```r
F_critico <- qf(0.95, 2, 203)

R2_r <- 0.03191
R2_ur <- 0.04059

F_est <- ((R2_ur - R2_r) / (1 - R2_ur)) * (203 / 2)

F_est
F_critico

if (F_est > F_critico) {
  "Se rechaza H0"
} else {
  "No se rechaza H0"
}
```

**Respuesta**

Se tiene que *F_est* es 0.91829353456812 y *F_critico* es 3.04037936353184, Como *F_est*<*F_critico no se rechaza la hipótesis nula al 5 %. Esto quiere decir que no existe evidencia estadística suficiente para concluir que las variables ventas_sobre_promedio y ventas_sobre_promedio_roe sean conjuntamente significativas en la explicación del salario.

4. (15 puntos) Volviendo a la regresión del salario en función de sales, roe y ros. Considere
dos grupos, Grupo 1: Aquellos clientes que sales es mayor al promedio, y Grupo 2: Aquellos
clientes que sales es menor al promedio. Realice un test de hipotesis para grupos al 5% de
significancia. Use el metodo que más le acomode.


```r
# Crear interacciones para comparar grupos
ceosal1$sales_D <- ceosal1$sales * ceosal1$ventas_sobre_promedio
ceosal1$roe_D   <- ceosal1$roe   * ceosal1$ventas_sobre_promedio
ceosal1$ros_D   <- ceosal1$ros   * ceosal1$ventas_sobre_promedio

modelo3 <- lm(salary ~ sales + roe + ros + ventas_sobre_promedio + sales_D + roe_D + ros_D,
              data = ceosal1)

summary(modelo3)

# Test F
F_critico <- qf(0.95, 4, 201)

R2_ur <- 0.06028
R2_r  <- 0.03191

F_est <- ((R2_ur - R2_r) / (1 - R2_ur)) * (201 / 4)

F_est
F_critico

if (F_est > F_critico) {
  "Se rechaza H0"
} else {
  "No se rechaza H0"
}
```

**Respuesta**

Como *F_est*<*F_critico no se rechaza la hipótesis nula al 5 %. Esto quiere decir que no existe evidencia estadística suficiente para afirmar que el modelo salarial difiera entre empresas con ventas altas y empresas con ventas bajas.

5. (15 puntos) Usando la regresión del salario en función de sales, roe y ros, se distinguen los
grupos mediante las dummies; indus, finance, consprod y utility, que corresponden al tipo de
empresa. Realice mediante test de Chow, con 5% de significancia, si los modelos para cada
grupo son iguales.

```r
ceosal1_indus    <- subset(ceosal1, indus == 1)
ceosal1_finance  <- subset(ceosal1, finance == 1)
ceosal1_consprod <- subset(ceosal1, consprod == 1)
ceosal1_utility  <- subset(ceosal1, utility == 1)

modelo_r <- lm(salary ~ sales + roe + ros, data = ceosal1)

modelo_indus    <- lm(salary ~ sales + roe + ros, data = ceosal1_indus)
modelo_finance  <- lm(salary ~ sales + roe + ros, data = ceosal1_finance)
modelo_consprod <- lm(salary ~ sales + roe + ros, data = ceosal1_consprod)
modelo_utility  <- lm(salary ~ sales + roe + ros, data = ceosal1_utility)


SSR_r <- sum(residuals(modelo_r)^2)

SSR_indus    <- sum(residuals(modelo_indus)^2)
SSR_finance  <- sum(residuals(modelo_finance)^2)
SSR_consprod <- sum(residuals(modelo_consprod)^2)
SSR_utility  <- sum(residuals(modelo_utility)^2)

SSR_ur <- SSR_indus + SSR_finance + SSR_consprod + SSR_utility

G  <- 4      # número de grupos
k  <- 3      # número de regresores (sales, roe, ros)
n  <- nrow(ceosal1)

gl1 <- (G - 1) * (k + 1)
gl2 <- n - G * (k + 1)

# F crítico
F_critico <- qf(0.95, gl1, gl2)

# Estadístico F de Chow
F_est <- ((SSR_r - SSR_ur) / gl1) / (SSR_ur / gl2)

F_est
F_critico

if (F_est > F_critico) {
  "Se rechaza H0"
} else {
  "No se rechaza H0"
}
```

Como *F_est*<*F_critico no se rechaza la hipótesis nula al 5 %. Por lo tanto, no existe evidencia estadística suficiente para afirmar que los modelos difieren entre los cuatro sectores.
